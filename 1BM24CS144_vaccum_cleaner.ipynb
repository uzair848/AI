```python
# ==========================================
# VACUUM CLEANER PROBLEM
# ==========================================

state = ['A', 'D', 'D']
cost = 0

print("Initial State:", state)
print("Cost:", cost)

if state[0] == 'A' and state[1] == 'D':
    state[1] = 'C'
    cost += 1

print("\nAction: Pick the dust")
print("State:", state)
print("Cost:", cost)

if state[0] == 'A':
    state[0] = 'B'
    cost += 1

print("\nAction: Move Right")
print("State:", state)
print("Cost:", cost)

if state[0] == 'B' and state[2] == 'D':
    state[2] = 'C'
    cost += 1

print("\nAction: Pick the dust")
print("State:", state)
print("Cost:", cost)

if state == ['B', 'C', 'C']:
    print("\nGoal State Reached!")
    print("Final State:", state)
    print("Total Cost:", cost)
else:
    print("\nGoal State Not Reached")


# ==========================================
# SIMPLE VACUUM AGENT
# ==========================================

def vacuum_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Right"
    elif location == "B":
        return "Left"


location = input("Enter location (A/B): ")
status = input("Enter status (Clean/Dirty): ")

action = vacuum_agent(location, status)

print("Location:", location)
print("Status:", status)
print("Action:", action)
```
