```python
import math


def alpha_beta(depth, node_index, is_maximizing, values, alpha, beta, max_depth):
    # Base case: reached a leaf node
    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        best = -math.inf

        # Recur for left and right children (binary tree)
        for i in range(2):
            val = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth,
            )
            best = max(best, val)
            alpha = max(alpha, best)

            # Alpha Beta Pruning condition
            if beta <= alpha:
                print(
                    f"Pruned branch at depth {depth + 1}, node index"
                    f" {node_index * 2 + 1}"
                )
                break
        return best

    else:
        best = math.inf

        # Recur for left and right children (binary tree)
        for i in range(2):
            val = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth,
            )
            best = min(best, val)
            beta = min(beta, best)

            # Alpha Beta Pruning condition
            if beta <= alpha:
                print(
                    f"Pruned branch at depth {depth + 1}, node index"
                    f" {node_index * 2 + 1}"
                )
                break
        return best


if __name__ == "__main__":
    # 1. Take tree depth as input
    max_depth = int(input("Enter the depth of the game tree (e.g., 3): "))
    num_leaves = 2**max_depth

    # 2. Take leaf node values as input
    print(f"Enter {num_leaves} leaf node values separated by spaces:")
    values = list(map(int, input().split()))

    # Validate input count
    if len(values) != num_leaves:
        print(
            f"Error: Expected exactly {num_leaves} values, but got"
            f" {len(values)}."
        )
    else:
        # 3. Ask who starts first
        turn = input(
            "Does the Maximizer start first? (y/n): "
        ).strip().lower()
        is_maximizing = turn == "y"

        # Run algorithm
        optimal_value = alpha_beta(
            0, 0, is_maximizing, values, -math.inf, math.inf, max_depth
        )
        print(f"\nThe optimal value is: {optimal_value}")
```
