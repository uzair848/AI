```python
def compute_cost(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbors(state):
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(tuple(neighbor))
    return neighbors

def format_state(state):
    return ", ".join(f"x{i} = {val}" for i, val in enumerate(state))

def hill_climbing(initial_state):
    current_state = initial_state

    while True:
        current_cost = compute_cost(current_state)
        print(f"\nNeighbours of {format_state(current_state)} (cost is {current_cost}.)")

        neighbors = get_neighbors(current_state)
        neighbor_costs = [compute_cost(nb) for nb in neighbors]

        min_cost = min(neighbor_costs)
        best_neighbor = None

        for nb, cost in zip(neighbors, neighbor_costs):
            chosen_label = ""
            if cost == min_cost and best_neighbor is None and cost < current_cost:
                best_neighbor = nb
                chosen_label = " Choose this neighbour."

            print(f" • {format_state(nb)} cost is {cost}.{chosen_label}")

        if best_neighbor is None:
            print(f"\n🛑 Local optimum or solution reached at: {current_state}")
            return current_state

        current_state = best_neighbor

def get_user_input():
    print("--- 4-Queens Hill Climbing Configuration ---")
    while True:
        try:
            user_in = input("Enter 4 space-separated row numbers (0 to 3): ").strip().split()
            state = tuple(int(x) for x in user_in)
            if len(state) != 4 or any(x < 0 or x > 3 for x in state):
                raise ValueError("Must provide exactly 4 values between 0 and 3.")
            return state
        except ValueError as e:
            print(f"Invalid input: {e}")

if __name__ == "__main__":
    initial = get_user_input()
    hill_climbing(initial)
```
